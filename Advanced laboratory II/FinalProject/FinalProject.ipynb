##Analisis de Imagenes Para Coeficientes de Cobertura (p)

```python
# ======================================
# CONFIGURACIÓN
# ======================================
def Imagenes(imagen):

  AREA_MIN = 50      # Área mínima en píxeles

# ======================================
# LEER IMAGEN
# ======================================

  img = cv2.imread(imagen, cv2.IMREAD_GRAYSCALE)

  if img is None:
    raise Exception("No se pudo abrir la imagen.")

# ======================================
# ASEGURAR IMAGEN BINARIA
# ======================================

  _, binaria = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

# ======================================
# LAS ISLAS DEBEN SER BLANCAS
# (OpenCV detecta mejor objetos blancos)
# ======================================

  binaria = cv2.bitwise_not(binaria)

# ======================================
# ENCONTRAR CONTORNOS
# ======================================

  contornos, _ = cv2.findContours(
    binaria,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
  )

# Imagen para dibujar resultados

  resultado = cv2.cvtColor(img,cv2.COLOR_GRAY2BGR)

  areas=[]
  perimetros=[]
  diametros=[]
  circularidades=[]
  centros=[]

# ======================================
# ANALIZAR CADA ISLA
# ======================================

  for i,c in enumerate(contornos):

    area = cv2.contourArea(c)

    if area < AREA_MIN:
        continue

    peri = cv2.arcLength(c,True)

    diam = np.sqrt(4*area/np.pi)

    circ = 4*np.pi*area/(peri**2)

    M = cv2.moments(c)

    if M["m00"]==0:
        continue

    cx = int(M["m10"]/M["m00"])
    cy = int(M["m01"]/M["m00"])

    centros.append([cx,cy])

    areas.append(area)
    perimetros.append(peri)
    diametros.append(diam)
    circularidades.append(circ)

    cv2.drawContours(resultado,[c],-1,(0,255,0),2)

    cv2.circle(resultado,(cx,cy),3,(0,0,255),-1)

    cv2.putText(resultado,
                str(len(areas)),
                (cx-8,cy+8),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.4,
                (255,0,0),
                1)

# ======================================
# COBERTURA
# ======================================

  pixeles_totales = binaria.shape[0]*binaria.shape[1]

  pixeles_islas = np.sum(binaria==255)

  p = pixeles_islas/pixeles_totales

# ======================================
# DISTANCIA AL VECINO MÁS CERCANO
# ======================================

  centros=np.array(centros)

  if len(centros)>1:

    D=cdist(centros,centros)

    np.fill_diagonal(D,np.inf)

    vecino=np.min(D,axis=1)

  else:

    vecino=[]

# ======================================
# RESULTADOS
# ======================================

  print("\n==========================")
  print("RESULTADOS")
  print("==========================")

  print(f"Número de islas: {len(areas)}")
  print(f"Coeficiente de cobertura: {p:.4f}")
  print(f"Área promedio: {np.mean(areas):.2f} px²")
  print(f"Perímetro promedio: {np.mean(perimetros):.2f} px")
  print(f"Diámetro equivalente promedio: {np.mean(diametros):.2f} px")
  print(f"Circularidad promedio: {np.mean(circularidades):.3f}")

  if len(vecino)>0:
    print(f"Distancia media entre centros: {np.mean(vecino):.2f} px")

# ======================================
# EXPORTAR CSV
# ======================================

  df = pd.DataFrame({
    "Área (px²)":areas,
    "Perímetro (px)":perimetros,
    "Diámetro equivalente (px)":diametros,
    "Circularidad":circularidades
  })

  df.to_csv("Microestructura.csv",index=False)

# ======================================
# HISTOGRAMA
# ======================================

  plt.figure(figsize=(6,4))
  plt.hist(areas,bins=20)
  plt.xlabel("Área (px²)")
  plt.ylabel("Frecuencia")
  plt.title("Distribución del área de las islas")
  plt.tight_layout()

# ======================================
# MOSTRAR
# ======================================

  plt.figure(figsize=(15,5))

  plt.subplot(131)
  plt.imshow(img,cmap="gray")
  plt.title("Original")
  plt.axis("off")

  plt.subplot(132)
  plt.imshow(binaria,cmap="gray")
  plt.title("Binaria")
  plt.axis("off")

  plt.subplot(133)
  plt.imshow(cv2.cvtColor(resultado,cv2.COLOR_BGR2RGB))
  plt.title("Islas detectadas")
  plt.axis("off")

  plt.tight_layout()
  plt.show()

  return
```

El area de cada isla metalica por pelıcula aumenta progresivamente desde la peĺıcula A hasta la pelıcula D. El incremento del tamano en las islas produce un aumento del  ́area metalica depositada y, en consecuencia, del coeficiente de cobertura de cada pelıcula. Se observa que el cambio en el  ́area de las pel ́ıculas ABC, se da en rangos relativamente pequẽnos comparado con el cambio C-D, hecho que se evidencia tambien en los valores del coeficiente de cobertura p

##R VS p

```python
p = [0.7874, 0.7954,  0.8033, 0.8546] #B,A,C,D
R = [0.0019, 0.0120, 0.0191, 0.4352]
```

```python
plt.figure(figsize=(8,5))

plt.plot(p, R,"--",  c = "#79C2F2" )
plt.plot(p, R,"o",  c = "#020A3D" )


plt.title("R Vs p")
plt.xlabel("Coeficiente de cobertura")
plt.ylabel("Resistencia promedio(TΩ)")
plt.grid(True)
plt.legend()
```

R vs p, evidencia que la resistencia promedio es directamente proporcional a el coeficiente de cobertura. Segun la teorıa que describe la conduccion eĺectrica en pelıculas delgadas discontinuas, en esta medicion se muestra que la conduccion electrica no esta dominada por el transporte entre islas metalicas, pues de ser ası, el resultado serıa opuesto.
